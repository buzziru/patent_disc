```python
def build_inputs(ex):
    fields = [
        ("명칭", ex["invention_title"]),
        ("IPC",  ex["ipc_main"]),
        ("요약", ex["abstract"]),
        ("청구항", ex["claims"]),
    ]
    text = " ".join(f"{k}: {v}" for k, v in fields if v)   # 빈 필드 skip, 개행/들여쓰기 없음
    return {"document_id": ex["document_id"], "input": text, "label_ids": ex["label_ids"]}
```
